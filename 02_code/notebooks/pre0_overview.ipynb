# Overview of the preCMGsim repo
Goal: The preCMGsim repo processes and converts a Petrel geomodel into CMG compatible formats, and sets up CMG dat files that fully define fluid flow and geomechanical coupled simulations. With inputs of realizations of reservoir properties (e.g., porosity and permeability) from files generated by a Petrel Uncertainty & Optimization workflow, this repo automatically sets up CMG dat files that will be run by CMG simulators. The implementation in this repo is the first part of a physics-based workflow for seismic hazard assessment for carbon capture and storage (CCS) and geothermal by integrating geomodeling, coupled flow-geomechanical simulation, fault slip modeling, global sensitivity analysis, and importance sampling. Below are high-level steps. Notebooks are organized in the following order.

## 1. Perform geomodeling in Petrel and create grid files for CMG simulations

## 2. Run a Petrel Uncertainty & Optimization workflow to generate geological realizations that contain finit/f0000 files, extract realizations (e.g., porosity and permeability) and convert them into CMG compatible format

## 3. Set up a CMG dat file template which contains all constant reservoir properties and simulation settings. In the template, set realization properties as variables. Finally, generate dat files by replacing variables with realizations (e.g., file paths to 3D properties, sampled values for flow and geomechanical properties)

## 4. Run simulations on these dat files in batches (e.g., batch of 100) on a computing cluster
