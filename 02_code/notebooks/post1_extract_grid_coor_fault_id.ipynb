# This notebook extracts average x, y, z coordinates and fault ID for each grid cell and generates a numpy array in the shape of (i, j, k, 4)

## Step 1: Export fgrid and fault ID (GRDECL) from Petrel 

Make sure these two files are exported using the same setting, including setting cell origin to be (I=0, J=max, K). This is because, in Petrel, i goes from left to right (west to east), j goes from top to bottom (north to south). In CMG, i is the same, but j goes from bottom to top (south to north).

## Step 2: Extract average x, y, z coordinates for each cell from a fgrid file

```python
from pathlib import Path
import sys

repo_root = Path.cwd().parent
sys.path.append(str(repo_root / "src"))

from fgrid2npy import fgrid2npy

################# Start of user inputs ########## 
fgrid_file_path = repo_root/'data'/'Vienna_geothermal'/'raw'/'JD_geothermal.fgrid' # path to the fgrid file
save_path = repo_root/'results'/'Vienna_geothermal'/'coor_fault'/'JD_geothermal_coor&fault' # path to save the processed numpy array
################## End of user inputs  ##########

coor_npy = fgrid2npy(
    fgrid_file_path = fgrid_file_path,
    save_dir = save_path
    )
```
