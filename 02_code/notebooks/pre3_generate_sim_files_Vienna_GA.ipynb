# This notebook generates CMG dat files for running CMG simulations

## Step 1: Set up a base CMG model
* Prepare a base CMG dat file and add it to data/dat_file_templates
* Note CMG requires initializing stress state using a reference block. For the JD_geothermal grid (139, 248, 23), reservoir is between k=5 and 20. Use block (50, 248, 5) as reference block for *STRESSGRAD calculation. Its grid top = 2365.807 m and bottom = 2383.534 m.

## Step 2: Sample uncertain parameters

* Note PORO/PERMX pairs are NOT sampled more than once (from file names instead of files)

```python
from pathlib import Path

notebook_path = Path.cwd()

from pre_parameter_sampling import latin_hypercube_sampling

################## User Inputs ############################## 
# Note: 1) stress gradients are effective ones (required by CMG) after subtracting 10; 
#       2) stress gradients are negative due to CMG DIR DOWN convention
param_bounds = {
    'KVKH'                      : [0.01,      0.5],
    'PRES_init_kPa'             : [24000,      25000],
    'FTM0'                       : [0.0,        0.5],
    'FTM1'                       : [0.0,        0.5],
    'HCAP_res_J/m3-C'     : [2.304e6,    3.117e6],
    'THCOND_res_J/m-day-C': [2.387e5, 2.920e5],
    'THEXPCOEF_1/C'             : [6.0e-6,    1.1e-5],
    'E_kPa'                     : [30e6,       35e6],
    'PR'                        : [0.2,        0.4],
    'SH_MPa/km'                 : [-8.68 * 1.05,  -8.68 * 0.95],
    'Sh_MPa/km'                 : [-5.52 * 1.05, -5.52 * 0.95],
    'Sv_MPa/km'                 : [-12.47 * 1.05,  -12.47 * 0.95],
    'SH_azi_deg'                : [158,          178],
}    # param_name : [lower, upper]
sampling_results = latin_hypercube_sampling(
    name_prefix = 'gen_000', # prefix for the output file name
    random_seed = 11, # random seed for Latin Hypercube sampling
    property_file_names_path = notebook_path/'data_other'/'260904_realization_property_filenames.csv', # path for the property file names
    output_file_path = notebook_path/'results'/'sim_dat_files', # path for the output file
    n_samples = 10,  # number of unique poro/permx pairs to sample (should be <= number of available pairs)
    param_names   = list(param_bounds.keys()),
    lower_bounds  = [b[0] for b in param_bounds.values()],
    upper_bounds  = [b[1] for b in param_bounds.values()],
    property_list = ['PORO','PERMX','TEMP'], # list of properties (CMG keywords) to sample from
    ref_block_top_depth = 2365.807,   # initial stress reference block, for the JD_Sula_2005_gmc grid, the reference block is (50, 1, 6) 
    ref_block_bottom_depth = 2383.534,
    show_results = True
)
################## End of User Inputs #######################
```

## Step 3: generate CMG dat files based on the sampled parameters

```python
import pandas as pd
from pathlib import Path
from pre_generate_dat_files import generate_dat_files

notebook_path = Path.cwd()
sim_dir = notebook_path/'results'/'sim_dat_files'

name_prefix = 'gen_000'   # prefix for the output file name
random_seed = 11          # random seed for Latin Hypercube sampling
case_col    = 'case'      # case-number column name expected by generate_dat_files

# 1) Well locations (all layouts): pro1_i -> PRO1_I to match the --PRO1_I-- placeholders
df_wells = pd.read_csv(notebook_path/'data_layout_gen'/f"{name_prefix}_to_simulate.csv")
df_wells = df_wells.rename(columns={c: c.upper() for c in df_wells.columns if c != 'layout_id'})
df_wells

# 2) Sampled variables 
df_params = pd.read_csv(sim_dir/f"{name_prefix}_sampled_params_seed{random_seed}.csv")
df_params.insert(0, 'param_id', range(1, len(df_params)+1))
df_params

# 3) Every layout x every parameter set, with continuous case numbers
df_all = df_wells.merge(df_params, how='cross')
df_all.insert(0, case_col, range(1, len(df_all)+1))
df_all.to_csv(sim_dir/f"{name_prefix}_cases_seed{random_seed}.csv", index=False)
print(f"{len(df_wells)} layouts x {len(df_params)} param sets = {len(df_all)} cases")
df_all

generate_dat_files(
    ################## User Inputs ##############################
    df_parameters = df_all,
    template_file_path = notebook_path/'data_other'/'JD_geothermal_base_model_GA.dat',
    save_folder_path = sim_dir/f"{name_prefix}_dat_files"
    ################## End of User Inputs #######################
)
```
