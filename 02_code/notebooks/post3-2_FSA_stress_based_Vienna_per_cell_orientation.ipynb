# This notebook performs fault slip analysis (FSA) based on principal stresses from CMG geomechanical simulations. 
- The following data are needed: (1) Grid cell coordinates, fault ID, dip angle and dip azimuth extracted from Petrel exported files, and (2) Principal stresses (STRESMXP, STRESMNP, STRESINT) extracted from CMG simulation results.
- For Vienna geothermal, normal faulting is used Sv > SH > Sh.
- **This version uses the per-cell fault orientation** from `JD_geothermal_coor&fault&dip.npy` (layer 4 = dip angle, layer 5 = dip azimuth) instead of one average strike/dip per fault. Because the faults are curved, this spreads the slip tendency over a range of values so slip nucleates progressively on the best-oriented patches rather than switching on across the whole fault at once.
- Outputs use `output_prefix` so they do not overwrite the uniform-orientation results.

# Step 1: Peform fault slip analysis

## Calculate and save the following (all are numpy arrays):  
1. Fault slip indicator (FSI: 0 no slip, 1 slip) in a folder: shape (n_is,n_js,n_ks,n_times) for each case
2. The ratio of shear stress to effective normal stress in a folder: shape (n_is,n_js,n_ks,n_times) for each case
3. Mean stress ratio in a file: shape (n_cases,n_faults,n_times)
4. Mean shear stress in a file: shape (n_cases,n_faults,n_times)
5. Total number of slipped cells in a file: shape (n_cases,n_faults,n_times)

```python
import numpy as np
import pandas as pd
from pathlib import Path
from tqdm import tqdm
import sys

# add src to path
repo_root = Path.cwd().parent
sys.path.append(str(repo_root / "src"))

from fault_slip_analysis import FSA_stress_based_per_cell

################## User Inputs ############################## 
name_prefix = '260805' # file name prefix
output_prefix = f'{name_prefix}_percell' # separate prefix so the uniform-orientation results are not overwritten
n_cases = 80 # number of simulation cases
n_faults = 2 # number of faults
n_times = 3 # number of simulation time steps, add the initial time step
fault_cohesion = 0 # fault cohesion in kPa
friction_coefficient = 0.6 # fault friction coefficient
n_is,n_js,n_ks = 139, 248, 23 # number of grid cells in the i,j,k directions
# paths to input data
parameters_csv_path = repo_root/'data'/'Vienna_geothermal'/'params_responses'/f'{name_prefix}_CMG_parameters.csv' # path to the parameters CSV file
principal_stresses_folder = repo_root/'results'/'Vienna_geothermal'/'principal_stresses'/f'{name_prefix}_sim_npy' # path to the folder containing the principal stress numpy files
coor_fault_npy_path = repo_root/'results'/'Vienna_geothermal'/'coor_fault'/'JD_geothermal_coor&fault&dip.npy' # coordinates, fault ID, dip angle and dip azimuth
# paths to output folder
fault_slip_analysis_folder = repo_root/'results'/'Vienna_geothermal'/'fault_slip_analysis'
################## End of User Inputs #######################

# load data
coor_fault = np.load(coor_fault_npy_path)
parameters = pd.read_csv(parameters_csv_path)

# per-cell fault geometry from the Petrel export (layer 4 = dip angle, layer 5 = dip azimuth)
# strike follows the right-hand rule: strike = dip_azimuth - 90
fault_masks   = {f: (coor_fault[:,:,:,3] == f) for f in range(n_faults)}
fault_dips    = {f: coor_fault[:,:,:,4][fault_masks[f]] for f in range(n_faults)}
fault_strikes = {f: (coor_fault[:,:,:,5][fault_masks[f]] - 90) % 360 for f in range(n_faults)}
for fault_id in range(n_faults):
    print(f'fault{fault_id}: {fault_masks[fault_id].sum()} cells, '
          f'dip {fault_dips[fault_id].mean():.2f} +/- {fault_dips[fault_id].std():.2f} deg')

# create save folders
stress_ratio_folder = fault_slip_analysis_folder/f'{output_prefix}_stress_ratio' # path to the folder to save stress ratio numpy files
stress_ratio_folder.mkdir(parents=True, exist_ok=True)
FSI_folder = fault_slip_analysis_folder/f'{output_prefix}_FSI' # path to the folder to save fault slip indicator numpy files
FSI_folder.mkdir(parents=True, exist_ok=True)

# initialize arrays to hold results for all cases, faults and time steps
mean_shear_stress   = np.full((n_cases,n_faults,n_times), np.nan) # array shape (n_cases,n_faults,n_times)
mean_stress_ratio   = np.full((n_cases,n_faults,n_times), np.nan) # array shape (n_cases,n_faults,n_times)
total_slipped_cells = np.full((n_cases,n_faults,n_times), np.nan) # array shape (n_cases,n_faults,n_times)

# run analysis
for case_num in tqdm(range(1,n_cases+1), desc='Running stress-based fault slip analysis'):
    # load principal stress arrays, make sure it fits to faulting regime
    Sv = np.load(principal_stresses_folder/f'case{case_num}_STRESMXP.npy')
    Sh = np.load(principal_stresses_folder/f'case{case_num}_STRESMNP.npy')
    SH = np.load(principal_stresses_folder/f'case{case_num}_STRESINT.npy')
    # extract the azimuth of the maximum horizontal stress from the parameter dataframe
    SH_azi = parameters.loc[parameters["case_name"] == f'case{case_num}','SH_azi_deg'].iloc[0]

    # per-cell results for this case, reset each case so nothing carries over
    stress_ratio_by_fault = np.full((n_is,n_js,n_ks,n_times), np.nan) # array shape (n_is,n_js,n_ks,n_times)
    FSI_by_fault          = np.full((n_is,n_js,n_ks,n_times), np.nan) # array shape (n_is,n_js,n_ks,n_times)

    for fault_id in range(n_faults):
        # mask a fault based on its id
        fault_id_mask = fault_masks[fault_id]

        # run fault slip analysis using the per-cell strike and dip of this fault.
        # the stresses are passed already masked, so sigma, tau and the slip
        # indicator come back with shape (n_cells,n_times)
        sigma, tau, fault_slip_indicator = FSA_stress_based_per_cell(
            SH = SH[fault_id_mask],
            Sh = Sh[fault_id_mask],
            Sv = Sv[fault_id_mask],
            SH_azi = SH_azi,
            fault_strike = fault_strikes[fault_id],
            fault_dip = fault_dips[fault_id],
            fault_cohesion = fault_cohesion,
            friction_coefficient = friction_coefficient
            )

        # calculate the mean shear stress for all cases, faults and time steps (note to covert kPa to MPa)
        mean_shear_stress[case_num-1,fault_id,:] = np.nanmean(tau/1000,axis=0) # array shape (n_cases,n_faults,n_times)

        # calculate stress ratio for each fault
        ratio = (tau - fault_cohesion) / sigma
        stress_ratio_by_fault[fault_id_mask] = ratio
        # calculate the mean stress ratio for all cases, faults and time steps
        mean_stress_ratio[case_num-1,fault_id,:] = np.nanmean(ratio,axis=0) # array shape (n_cases,n_faults,n_times)

        # extract fault slip indicator for each fault 
        FSI_by_fault[fault_id_mask] = fault_slip_indicator # array shape (n_is,n_js,n_ks,n_times)

        # calculate the total number of slipped cells for all cases, faults and time steps
        total_slipped_cells[case_num-1,fault_id,:] = np.nansum(fault_slip_indicator,axis=0) # array shape (n_cases,n_faults,n_times)

    # save stress ratio for each case
    np.save(stress_ratio_folder/f'case{case_num}_stress_ratio.npy', stress_ratio_by_fault)
    # save fault slip indicator for each case
    np.save(FSI_folder/f'case{case_num}_FSI.npy', FSI_by_fault)

# save the mean shear stress
np.save(fault_slip_analysis_folder/f'{output_prefix}_mean_shear_stress.npy',mean_shear_stress)
# save the mean stress ratio
np.save(fault_slip_analysis_folder/f'{output_prefix}_mean_stress_ratio.npy',mean_stress_ratio)
# save the total number of slipped cells
np.save(fault_slip_analysis_folder/f'{output_prefix}_total_slipped_cells.npy',total_slipped_cells)

print("\nFinished analyzing all cases.")
```

# Step 2: Calculate the (number of cases with fault slip / total cases)

```python
import numpy as np
import pandas as pd
from pathlib import Path

repo_root = Path.cwd().parent

################## User Inputs ############################## 
name_prefix = '260716_test' # file name prefix
total_slipped_cells_npy_path = repo_root/'results'/'Vienna_geothermal'/'fault_slip_analysis'/f'{name_prefix}_total_slipped_cells.npy' # path to the total slipped cells numpy file
save_path = repo_root/'results'/'Vienna_geothermal'/'fault_slip_analysis' # path to save results
################## End of User Inputs #######################

total_slipped_cells = np.load(total_slipped_cells_npy_path)
no_slip_count = np.sum(total_slipped_cells == 0, axis=0)
slip_probability = 1 - no_slip_count/total_slipped_cells.shape[0]
# np.savetxt(base_path/'data'/f'{save_file_prefix}_FSA_probability.csv',slip_probability,delimiter=",",fmt="%.4f")
# Save as CSV
df = pd.DataFrame(
    slip_probability,
    columns=[f"time_{t}" for t in range(n_times)]
)
df.insert(0, "fault_id", range(n_faults))

df.to_csv(save_path/f'{name_prefix}_slip_probability.csv', index=False, float_format="%.4f")
df
```
